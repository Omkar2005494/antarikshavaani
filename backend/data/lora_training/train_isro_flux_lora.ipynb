### Step 4: Download Your Trained Model File to Your Mac

```python
from google.colab import files
import os

weight_file = '/content/isro_lora_output/pytorch_lora_weights.safetensors'
if os.path.exists(weight_file):
    print('⬇️ Downloading isro_spacecraft_lora.safetensors to your computer...')
    files.download(weight_file)
else:
    # Fallback to any safetensors file in output
    for root, dirs, f_list in os.walk('/content/isro_lora_output'):
        for f in f_list:
            if f.endswith('.safetensors'):
                files.download(os.path.join(root, f))
                break
```
